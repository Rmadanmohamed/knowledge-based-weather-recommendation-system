```python
#  Install required libraries
!pip install networkx matplotlib pandas scikit-learn -q

#  Import libraries
import networkx as nx
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import LabelEncoder

#  Load knowledge graph data
data_dir = "/content/kg_project_data"
train_path = f"{data_dir}/train.tsv"

# Read the triples (head, relation, tail)
df = pd.read_csv(train_path, sep="\t", header=None, names=["head", "relation", "tail"])
# Build the graph  # This line was incorrectly indented
G = nx.Graph()
for _, row in df.iterrows():
    G.add_node(row["head"])
    G.add_node(row["tail"])
    G.add_edge(row["head"], row["tail"], label=row["relation"])

#  Label Encoding for cities
label_encoder = LabelEncoder()
cities = pd.concat([df["head"], df["tail"]]).unique()
label_encoder.fit(cities)

#  Prepare data
relation_map = {'isCapitalOf': 0, 'locatedIn': 1, 'hasAirport': 2, 'isPartOf': 3}
X = []
y = []

for _, row in df.iterrows():
    if row["relation"] in relation_map:
        X.append([row["head"], row["tail"]])
        y.append(relation_map[row["relation"]])

X_encoded = []
for city1, city2 in X:
    city1_encoded = label_encoder.transform([city1])[0]
    city2_encoded = label_encoder.transform([city2])[0]
    X_encoded.append([city1_encoded, city2_encoded])

#  Split & Train
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=42)
model = DecisionTreeClassifier()
model.fit(X_train, y_train)

#  Accuracy
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Model Accuracy: {accuracy * 100:.2f}%")

city_continent_map = {
    "Cairo": "Africa", "Alexandria": "Africa", "London": "Europe", "Paris": "Europe", "Berlin": "Europe",
    "Madrid": "Europe", "Rome": "Europe", "Nairobi": "Africa", "Cape Town": "Africa",
    "Buenos Aires": "South America", "Sao Paulo": "South America", "New York": "North America",
    "Los Angeles": "North America", "Toronto": "North America", "Tokyo": "Asia", "Beijing": "Asia",
    "Mumbai": "Asia", "Sydney": "Australia", "Melbourne": "Australia", "Moscow": "Europe", "Istanbul": "Asia",
    "Dubai": "Asia", "Johannesburg": "Africa", "Lagos": "Africa", "Seoul": "Asia", "Bangkok": "Asia",
    "Lima": "South America", "Caracas": "South America"
}


def predict_relation(city1, city2):
    if city1 not in label_encoder.classes_ or city2 not in label_encoder.classes_:
        return " One or both cities are not recognized by the model."

    city1_encoded = label_encoder.transform([city1])[0]
    city2_encoded = label_encoder.transform([city2])[0]
    prediction = model.predict([[city1_encoded, city2_encoded]])[0]
    relation = [k for k, v in relation_map.items() if v == prediction][0]
    return f"The predicted relation between '{city1}' and '{city2}' is: **{relation}**"


def get_common_continent(city1, city2):
    continent1 = city_continent_map.get(city1)
    continent2 = city_continent_map.get(city2)
    if continent1 and continent2:
        if continent1 == continent2:
            return f" Both cities are located in the continent: **{continent1}**."
        else:
            return f" Different continents: {city1} is in **{continent1}**, {city2} is in **{continent2}**."
    else:
        return " One or both cities do not have known continent data."


city1 = input("Enter the first city: ")
city2 = input("Enter the second city: ")


print(predict_relation(city1, city2))
print(get_common_continent(city1, city2))
```

```python
import pandas as pd
import spacy

# Load spacy model
nlp = spacy.load("en_core_web_sm")

# Load the dataset
df = pd.read_csv('/content/world_countries_dataset.csv')

# Normalize data
df['Country'] = df['Country'].str.lower()
df['Language'] = df['Language'].str.lower()
df['Currency'] = df['Currency'].str.lower()

def get_entity(doc):
    entities = {"COUNTRY": [], "LANGUAGE": [], "CURRENCY": []}
    for token in doc:
        if token.text in df['Country'].values:
            entities["COUNTRY"].append(token.text)
        elif token.text in df['Language'].values:
            entities["LANGUAGE"].append(token.text)
        elif token.text in df['Currency'].values:
            entities["CURRENCY"].append(token.text)
    return entities

def search(entities):
    if entities["COUNTRY"]:
        for country in entities["COUNTRY"]:
            result = df[df['Country'] == country]
            if not result.empty:
                print("\n--- Country Information ---")
                print(result.to_string(index=False))
            else:
                print(f"No information found for {country}")

    elif entities["LANGUAGE"]:
        for language in entities["LANGUAGE"]:
            result = df[df['Language'].str.contains(language)]
            if not result.empty:
                print("\n--- Countries speaking this language ---")
                print(result[['Country', 'Language']].to_string(index=False))
            else:
                print(f"No countries found speaking {language}")

    elif entities["CURRENCY"]:
        for currency in entities["CURRENCY"]:
            result = df[df['Currency'].str.contains(currency)]
            if not result.empty:
                print("\n--- Countries using this currency ---")
                print(result[['Country', 'Currency']].to_string(index=False))
            else:
                print(f"No countries found using {currency}")
    else:
        print("Sorry, couldn't understand your question.")

def main():

    while True:
        question = input("\nAsk your question (or type 'exit' to quit): ").lower().strip()

        if question == 'exit':
            print("Goodbye! ")
            break

        doc = nlp(question)
        entities = get_entity(doc)
        search(entities)

if __name__ == "__main__":
    main()
```

```python
import spacy
import pandas as pd
import sys

# Load spaCy model (Deep Learning based NLP)
nlp = spacy.load("en_core_web_sm")

# Add the directory path of your data_tuples module (NOT the file itself)
sys.path.append('/content/country_data_tuples.py')

# Import your tuples from the file
from country_data_tuples import data_tuples

# Convert tuples to DataFrame
columns = ['Country', 'Language', 'Population', 'Currency', 'Region']
df = pd.DataFrame(data_tuples, columns=columns)

# Normalize
df['Country'] = df['Country'].str.lower()
df['Language'] = df['Language'].str.lower()
df['Currency'] = df['Currency'].str.lower()

# Entity recognition from question
def get_entity(doc):
    entities = {"COUNTRY": [], "LANGUAGE": [], "CURRENCY": []}
    for token in doc:
        word = token.text.lower()
        if word in df['Country'].values:
            entities["COUNTRY"].append(word)
        elif word in df['Language'].values:
            entities["LANGUAGE"].append(word)
        elif word in df['Currency'].values:
            entities["CURRENCY"].append(word)
    return entities

# Search and respond like a chatbot
def search(entities):
    if entities["COUNTRY"]:
        for country in entities["COUNTRY"]:
            result = df[df['Country'] == country]
            if not result.empty:
                row = result.iloc[0]
                return f""" Here’s what I found about {row['Country'].title()}:
- Language: {row['Language'].title()}
- Population: {row['Population']:,}
- Currency: {row['Currency'].title()}
- Region: {row['Region'].title()}"""
            else:
                return f" Sorry, I couldn't find information about {country.title()}."

    elif entities["LANGUAGE"]:
        language = entities["LANGUAGE"][0]
        matches = df[df['Language'].str.contains(language)]
        if not matches.empty:
            countries = ", ".join(matches['Country'].str.title())
            return f" The language *{language.title()}* is spoken in: {countries}."
        else:
            return f" I couldn't find any countries that speak {language.title()}."

    elif entities["CURRENCY"]:
        currency = entities["CURRENCY"][0]
        matches = df[df['Currency'].str.contains(currency)]
        if not matches.empty:
            countries = ", ".join(matches['Country'].str.title())
            return f" The currency *{currency.title()}* is used in: {countries}."
        else:
            return f" No countries found using the {currency.title()}."

    else:
        return " I'm not sure I understood that. Try asking about a country, language, or currency."

# Main chatbot loop
def main():
   import gradio as gr

# Function to be used by the Gradio interface
def chatbot_response(question):
    doc = nlp(question.lower().strip())
    entities = get_entity(doc)
    return search(entities)

# Create the interface
interface = gr.Interface(
    fn=chatbot_response,
    inputs=gr.Textbox(lines=2, placeholder="Ask me about a country, language, or currency..."),
    outputs="text",
    title="🌍 Country Info Chatbot",
    description="Ask me about any country, language, or currency. Type naturally, like: 'Tell me about Egypt' or 'Which countries use the euro?'"
)

# Launch the interface
interface.launch()
```
